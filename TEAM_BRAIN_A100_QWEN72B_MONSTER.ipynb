```python
# 5. PYTORCH / TRANSFORMERS 72B MONSTER TEST
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

print("🚀 TESTE QWEN 2.5 72B AUF A100 (80GB VRAM)...", flush=True)

try:
    print(f"PyTorch Version: {torch.__version__}")
    print(f"CUDA Verfügbar: {torch.cuda.is_available()}")
    print(f"GPU: {torch.cuda.get_device_name(0)}")

    # 1. 72B AWQ Modell auswählen
    model_id = "Qwen/Qwen2.5-72B-Instruct-AWQ"
    print(f"\n⏳ Lade {model_id} via Transformers in die A100...")

    tokenizer = AutoTokenizer.from_pretrained(model_id)
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        torch_dtype=torch.float16,
        device_map="auto"
    )

    vram_used = round(torch.cuda.memory_allocated() / (1024**3), 2)
    print(f"\n✅ ERFOLG! 72B Modell im A100 VRAM geladen: {vram_used} GB belegt!")

    # 2. Test-Generierung mit 72B
    print("\n🧠 Generiere echte 72B-Antwort...")
    inputs = tokenizer("Schreibe ein kurzes Begrüßungs-Statement für Team Brain:", return_tensors="pt").to("cuda")
    outputs = model.generate(**inputs, max_new_tokens=50)
    antwort = tokenizer.decode(outputs[0], skip_special_tokens=True)

    print("\n" + "=" * 40)
    print("🔥 ANTWORT VON QWEN 2.5 72B:")
    print(antwort)
    print("=" * 40)

except Exception as e:
    print(f"\n❌ FEHLER: {e}")
```

```python
# 🦅 DIREKTER 72B GROSSAUFTRAG: TUCSON MOBILITY ENGINE (BENZIN / TANKERKÖNIG)
# =========================================================================
import time

prompt = """Du bist Senior Automotive Software Engineer bei Team Brain.
Schreibe ein vollständiges, produktionsreifes Python-Modul 'enterprise_mobility_engine.py' für den Hyundai Tucson (Benziner / 1.6 T-GDI).

Anforderungen:
1. Tankerkönig API Integration:
   - Live-Abfrage der günstigsten Tankstellen im Umkreis für E5, E10 und Super Plus.
   - Keine EV/Elektro-Ladefunktionen (reiner Verbrenner / Benzin).
2. Routing & Fuel-Cost Optimizer:
   - Berechne Routenkosten basierend auf Durchschnittsverbrauch (z.B. 7.2L/100km) und Live-Spritpreisen.
   - Finde die wirtschaftlichste Tankstelle entlang einer Route.
3. 100% Typsicherheit mit Pydantic / Dataclasses, sauberes Logging und robuste Fehlerbehandlung.
4. Gib nur den reinen, vollständigen Python-Code aus (ohne Auslassungen oder Platzhalter).
"""

print("🧠 Qwen 2.5 72B rechnet und generiert den Code (dauert ca. 30-60 Sekunden)...", flush=True)
start = time.time()

inputs = tokenizer(prompt, return_tensors="pt").to("cuda:0")

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=3000,
        temperature=0.2,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

code_out = tokenizer.decode(outputs[0], skip_special_tokens=True)
dauer = round(time.time() - start, 1)

# Datei direkt abspeichern
with open("/content/enterprise_mobility_engine.py", "w") as f:
    f.write(code_out)

print(f"\n✅ FERTIG! 72B hat den Code in {dauer}s generiert und unter /content/enterprise_mobility_engine.py gespeichert!")
print("=" * 60)
print(code_out[:1500] + "\n\n... [VOLLSTÄNDIGER CODE GESPEICHERT] ...")
print("=" * 60)
```

```python
# 🦅 TEAM BRAIN A100 GOOGLE-DRIVE TASK WORKER (100% TUNNEL-FREI)
# ===============================================================
import os, time, json, glob
from google.colab import drive

# 1. Drive sicher mounten
if not os.path.exists('/content/drive/MyDrive'):
    drive.mount('/content/drive', force_remount=False)

inbox_dir = "/content/drive/MyDrive/A100_Inbox"
outbox_dir = "/content/drive/MyDrive/A100_Outbox"
os.makedirs(inbox_dir, exist_ok=True)
os.makedirs(outbox_dir, exist_ok=True)

print("\n" + "=" * 60)
print("🦅 A100 QWEN 72B WORKER LAUSCHT AUF GOOGLE DRIVE!")
print(f"📥 Inbox:  {inbox_dir}")
print(f"📤 Outbox: {outbox_dir}")
print("=" * 60 + "\n")

# 2. Warteschleife auf Aufträge von Andy
while True:
    task_files = glob.glob(f"{inbox_dir}/*.json")
    if task_files:
        task_path = task_files[0]
        try:
            with open(task_path, "r", encoding="utf-8") as f:
                task = json.load(f)

            task_id = task.get("id", f"task_{int(time.time())}")
            prompt = task.get("prompt", "")
            max_tok = task.get("max_tokens", 3000)

            print(f"🧠 [A100 72B] Starte Auftrag: {task_id}...", flush=True)
            t_start = time.time()

            inputs = tokenizer(prompt, return_tensors="pt").to("cuda:0")
            with torch.no_grad():
                outputs = model.generate(
                    **inputs,
                    max_new_tokens=max_tok,
                    temperature=0.2,
                    do_sample=True,
                    pad_token_id=tokenizer.eos_token_id
                )
            result_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
            dauer = round(time.time() - t_start, 1)

            # Ergebnis in Outbox schreiben
            out_path = f"{outbox_dir}/{task_id}_result.txt"
            with open(out_path, "w", encoding="utf-8") as f:
                f.write(result_text)

            # Erledigten Auftrag aus Inbox löschen
            os.remove(task_path)
            print(f"✅ [A100 72B] Auftrag {task_id} fertig in {dauer}s! ➔ Gespeichert in Outbox.", flush=True)

        except Exception as e:
            print(f"❌ Fehler bei Auftrag {task_path}: {e}", flush=True)
            time.sleep(2)

    time.sleep(2)
```
