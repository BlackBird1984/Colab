```python
# 5. PYTORCH / TRANSFORMERS 72B MONSTER TEST
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer

print("🚀 TESTE QWEN 2.5 72B AUF A100 (80GB VRAM)...", flush=True)

try:
    print(f"PyTorch Version: {torch.__version__}")
    print(f"CUDA Verfügbar: {torch.cuda.is_available()}")
    print(f"GPU: {torch.cuda.get_device_name(0)}")

    # 1. 72B AWQ Modell auswählen
    model_id = "Qwen/Qwen2.5-72B-Instruct-AWQ"
    print(f"\n⏳ Lade {model_id} via Transformers in die A100...")

    tokenizer = AutoTokenizer.from_pretrained(model_id)
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        torch_dtype=torch.float16,
        device_map="auto"
    )

    vram_used = round(torch.cuda.memory_allocated() / (1024**3), 2)
    print(f"\n✅ ERFOLG! 72B Modell im A100 VRAM geladen: {vram_used} GB belegt!")

    # 2. Test-Generierung mit 72B
    print("\n🧠 Generiere echte 72B-Antwort...")
    inputs = tokenizer("Schreibe ein kurzes Begrüßungs-Statement für Team Brain:", return_tensors="pt").to("cuda")
    outputs = model.generate(**inputs, max_new_tokens=50)
    antwort = tokenizer.decode(outputs[0], skip_special_tokens=True)

    print("\n" + "=" * 40)
    print("🔥 ANTWORT VON QWEN 2.5 72B:")
    print(antwort)
    print("=" * 40)

except Exception as e:
    print(f"\n❌ FEHLER: {e}")
```

```python
# 🦅 TEAM BRAIN A100 GOOGLE-DRIVE TASK WORKER (100% TUNNEL-FREI)
import os, time, json, glob, torch
from transformers import AutoModelForCausalLM, AutoTokenizer

INBOX_DIR = "/content/gdrive/MyDrive/A100_Inbox"
OUTBOX_DIR = "/content/gdrive/MyDrive/A100_Outbox"
os.makedirs(INBOX_DIR, exist_ok=True)
os.makedirs(OUTBOX_DIR, exist_ok=True)

def extract_prompt(data):
    if not isinstance(data, dict):
        return None
    # Parser-Priorität: 1. prompt, 2. instruction (Kompatibilitätsfeld), 3. task
    prompt_val = data.get("prompt") or data.get("instruction") or data.get("task")
    if not prompt_val and "task_card" in data:
        tc = data["task_card"]
        if isinstance(tc, dict):
            prompt_val = tc.get("prompt") or tc.get("instruction") or tc.get("goal")
    if not prompt_val and "instructions" in data:
        ins = data["instructions"]
        if isinstance(ins, dict):
            prompt_val = ins.get("prompt") or ins.get("goal")
        elif isinstance(ins, str):
            prompt_val = ins
    # Ablehnung von leeren oder fehlenden Eingabetexten vor der Inferenz
    if prompt_val and isinstance(prompt_val, str) and prompt_val.strip():
        return prompt_val.strip()
    return None

print("🦅 TEAM BRAIN A100 GOOGLE-DRIVE TASK WORKER (100% TUNNEL-FREI) LAUSCHT AUF DRIVE!")
print(f"📥 Inbox:  {INBOX_DIR}")
print(f"📤 Outbox: {OUTBOX_DIR}")

while True:
    task_files = sorted(glob.glob(os.path.join(INBOX_DIR, "*.json")))
    for tf in task_files:
        task_id = os.path.basename(tf).replace(".json", "")
        out_path = os.path.join(OUTBOX_DIR, f"{task_id}_result.json")

        # Idempotenz-Sperre: Wenn bereits ein Ergebnis in der Outbox existiert, überspringe
        if os.path.exists(out_path):
            print(f"⏩ [A100 72B] Auftrag {task_id} bereits verarbeitet. Überspringe...")
            continue

        print(f"🧠 [A100 72B] Verarbeite Auftrag: {task_id}...")
        try:
            with open(tf, "r", encoding="utf-8") as f:
                data = json.load(f)
            prompt_text = extract_prompt(data)
            if not prompt_text:
                err_msg = f"❌ FEHLER: Kein gültiger Prompt/Instruction Text in {task_id}.json gefunden oder Text leer."
                print(err_msg)
                with open(out_path, "w", encoding="utf-8") as out_f:
                    json.dump({"task_id": task_id, "status": "REJECTED_EMPTY_PROMPT", "error": err_msg}, out_f, indent=2)
                continue

            # 72B Inferenz mit geladenem Tokenizer & Modell
            inputs = tokenizer(prompt_text, return_tensors="pt").to("cuda" if torch.cuda.is_available() else "cpu")
            with torch.no_grad():
                outputs = model.generate(**inputs, max_new_tokens=2048, do_sample=True, temperature=0.7)
            response_text = tokenizer.decode(outputs[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)

            with open(out_path, "w", encoding="utf-8") as out_f:
                json.dump({
                    "task_id": task_id,
                    "status": "SUCCESS",
                    "prompt_processed": prompt_text,
                    "response": response_text
                }, out_f, indent=2)
            print(f"✅ [A100 72B] Auftrag {task_id} erfolgreich verarbeitet!")
        except Exception as e:
            print(f"❌ Fehler bei Auftrag {tf}: {e}")
    time.sleep(3)
```
