# 🦅 BlackBird & Team Brain — Google Colab Starter

Dieses Notebook verbindet dein Colab-Setup direkt mit Google Gemini, Search Grounding und den Team Brain Agenten-Profilen.

### 1. ⚙️ Setup & Bibliotheken installieren

```python
# Neueste Google GenAI Bibliothek installieren
!pip install -q -U "google-genai"

import os
from google.colab import userdata
from google import genai

# Hole den Gemini API-Key aus den Colab Secrets (Schluessel-Symbol links)
try:
    api_key = userdata.get("GOOGLE_API_KEY") or userdata.get("GEMINI_API_KEY")
except Exception:
    api_key = os.environ.get("GEMINI_API_KEY")

if not api_key:
    print("⚠️ Bitte hinterlege deinen GEMINI_API_KEY links beim Schluessel-Symbol in Secrets!")
else:
    client = genai.Client(api_key=api_key)
    print("✅ Gemini Client erfolgreich initialisiert!")
```

### 2. ⚖️ Testlauf: Mabel (Lead Business-Architektur & Notare)

```python
# Mabel Prompt Test mit Google Gemini 3.8 Flash / 3.7 Flash
prompt = "Hallo Mabel! Erkläre mir kurz die Strategie für Notarhelden und worauf Notare besonders achten."

mabel_instruction = """
Du bist Mabel, die Lead Business-Architektin von Team Brain.
Tonalitaet: Locker, smart, empathisch auf Augenhoehe. Bei Bedarf Fachanwaeltin fuer IT- und Notarrecht.
Schwerpunkte: § 203 StGB, Notarvertraege, BNotK, Vertrauenspsychologie.
"""

response = client.models.generate_content(
    model="gemini-2.5-flash", # oder gemini-3.8-flash
    contents=prompt,
    config={
        "system_instruction": mabel_instruction,
    }
)

print("=== MABELS ANTWORT ===\n")
print(response.text)
```
