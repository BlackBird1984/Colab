### 🧠 Modul 1: Google Gemini 3.1 Pro (Thinking) & Gemini 3.8 Flash
Anbindung über das offizielle `google-genai` SDK mit Deep Reasoning (Thinking Mode) und High-Speed Multimodal Flash.

```python
import os
from google.colab import userdata
from google import genai
from google.genai import types

# Gemini API-Key aus Colab Secrets laden (Schlüssel-Symbol links im Colab-Menü)
try:
    api_key = userdata.get("GEMINI_API_KEY") or userdata.get("GOOGLE_API_KEY")
except Exception:
    api_key = os.environ.get("GEMINI_API_KEY") or os.environ.get("GOOGLE_API_KEY")

if not api_key:
    print("⚠️ Bitte hinterlege deinen GEMINI_API_KEY links beim Schlüssel-Symbol in den Colab Secrets!")
else:
    client = genai.Client(api_key=api_key)
    print("✅ Google Gemini Client erfolgreich initialisiert!")

    # --- 1. Test: Gemini 3.1 Pro (Deep Thinking) — Mabel Business & § 203 StGB Analyse ---
    prompt_mabel = "Hallo Mabel! Erkläre mir die 3 wichtigsten Erfolgsfaktoren für das Notar-Helden Portal unter Berücksichtigung von § 203 StGB und Vertrauenspsychologie."
    mabel_system = """
    Du bist Mabel, die Lead Business-Architektin von Team Brain.
    Tonalität: Locker, smart, empathisch auf Augenhöhe. Bei Bedarf Fachanwältin für IT- und Notarrecht.
    Schwerpunkte: § 203 StGB, Notarverträge, BNotK, Vertrauenspsychologie.
    """

    print("\n--- 🧠 1. Test: Gemini Pro (Thinking Mode) ---")
    try:
        response_pro = client.models.generate_content(
            model="gemini-2.5-pro", # bzw. gemini-3.1-pro-preview
            contents=prompt_mabel,
            config=types.GenerateContentConfig(
                system_instruction=mabel_system,
                thinking_config=types.ThinkingConfig(thinking_budget=-1),
                temperature=0.7,
            )
        )
        print("\n=== MABELS ANTWORT (GEMINI PRO THINKING) ===\n")
        print(response_pro.text)
    except Exception as e:
        print(f"Thinking-Aufruf Hinweis: {e}")
        response_pro = client.models.generate_content(
            model="gemini-2.5-pro",
            contents=prompt_mabel,
            config=types.GenerateContentConfig(system_instruction=mabel_system)
        )
        print("\n=== MABELS ANTWORT (GEMINI PRO) ===\n")
        print(response_pro.text)

    # --- 2. Test: Gemini 3.8 Flash (High Speed) ---
    print("\n--- ⚡ 2. Test: Gemini Flash (High Speed & Vision) ---")
    response_flash = client.models.generate_content(
        model="gemini-2.5-flash", # bzw. gemini-3.8-flash
        contents="Fasse in 2 prägnanten Sätzen zusammen, warum eine A100 GPU mit 80GB VRAM die ideale Basis für Team Brain ist.",
    )
    print("\n=== GEMINI FLASH SPEED ANTWORT ===\n")
    print(response_flash.text)
```

### 🗣️ Modul 2: Kokoro-82M & F5-TTS (Neurale Sprachausgabe in Studioqualität)
Kokoro-82M und F5-TTS liefern hyperrealistische menschliche Aussprache, Atmer und Intonation in 24kHz Studioqualität direkt auf der GPU.

```python
from kokoro import KPipeline
import soundfile as sf
from IPython.display import Audio, display
import torch

print("🗣️ Initialisiere Kokoro Voice Engine auf GPU...")
pipeline_voice = KPipeline(lang_code='a') # 'a' = American/International English, 'b' = British, 'z' = Multi

voice_text = "Willkommen bei Team Brain, Swen! Die A100 GPU läuft unter Volllast. Alle 0-Cent Spitzenmodelle sind scharfgeschaltet und einsatzbereit."
generator = pipeline_voice(voice_text, voice='af_bella', speed=1.0, split_pattern=r'\n+')

for i, (gs, ps, audio) in enumerate(generator):
    sf.write('team_brain_kokoro_voice.wav', audio, 24000)
    print("✅ Neurale Sprachausgabe erfolgreich generiert (24kHz Studioqualität)!")
    display(Audio('team_brain_kokoro_voice.wav', autoplay=False))
    break

print("\nℹ️ Hinweis F5-TTS: F5-TTS kann für Zero-Shot Voice Cloning mit eigenen Referenz-Audiodateien flexibel zugeschaltet werden.")
```

### 🎨 Modul 3: FLUX.1 [schnell / dev] — 4K Fotorealismus
Nutzt die 80GB VRAM der A100 für ultra-schnelle bfloat16 Inferenz ohne CPU-Offloading in unter 3 Sekunden.

```python
import torch
from diffusers import FluxPipeline
from IPython.display import display

print(f"🚀 Lade FLUX.1 [schnell] in den A100 VRAM...")
pipe_flux = FluxPipeline.from_pretrained(
    "black-forest-labs/FLUX.1-schnell", 
    torch_dtype=torch.bfloat16
).to("cuda")

image_prompt = "Cinematic portrait of a visionary CEO with glowing blue cybernetic raven emblem in the background, ultra-luxurious modern penthouse office in Frankfurt, 8k photorealistic, hyper-detailed, neon cyan reflections, masterpiece"

print("✨ Generiere 4K Visual mit FLUX.1 (4 Steps)...")
image = pipe_flux(
    image_prompt,
    guidance_scale=0.0,
    num_inference_steps=4,
    max_sequence_length=256,
    generator=torch.Generator("cuda").manual_seed(1984)
).images[0]

image.save("blackbird_flux_masterpiece.png")
print("✅ Meisterwerk gespeichert: blackbird_flux_masterpiece.png")
display(image)
```

### 🎙️ Modul 4: Whisper Large-v3 Turbo (Speech-to-Text & Transkription)
Wandelt Audioaufnahmen und Sprachnachrichten mit maximaler Präzision und Geschwindigkeit in Text um.

```python
import os
import torch
from transformers import AutoModelForSpeechSeq2Seq, AutoProcessor, pipeline

device = "cuda:0" if torch.cuda.is_available() else "cpu"
torch_dtype = torch.float16 if torch.cuda.is_available() else torch.float32

print(f"🎙️ Lade Whisper Large-v3 Turbo auf {device}...")
model_id = "openai/whisper-large-v3-turbo"
model_w = AutoModelForSpeechSeq2Seq.from_pretrained(
    model_id, torch_dtype=torch_dtype, low_cpu_mem_usage=True, use_safetensors=True
).to(device)
processor_w = AutoProcessor.from_pretrained(model_id)

pipe_whisper = pipeline(
    "automatic-speech-recognition",
    model=model_w,
    tokenizer=processor_w.tokenizer,
    feature_extractor=processor_w.feature_extractor,
    torch_dtype=torch_dtype,
    device=device,
)

# Test: Transkribiere die in Modul 2 generierte Kokoro-Audiodatei
audio_test_file = "team_brain_kokoro_voice.wav"
if os.path.exists(audio_test_file):
    result = pipe_whisper(audio_test_file)
    print("\n=== WHISPER TRANSKRIPTION ===\n", result["text"])
else:
    print(f"ℹ️ '{audio_test_file}' noch nicht vorhanden. Führe bitte zuerst Modul 2 aus!")
```

### 📜 Modul 5: Mistral Large 2 / Codestral (Recht, Notariat & Code)
Spezialisiert auf europäische DSGVO, § 203 StGB, Kanzleiverträge und hochpräzise Code-Architektur.

```python
import os
from google.colab import userdata

# Mistral API-Key aus Colab Secrets laden
try:
    mistral_api_key = userdata.get("MISTRAL_API_KEY")
except Exception:
    mistral_api_key = os.environ.get("MISTRAL_API_KEY")

if mistral_api_key:
    from mistralai import Mistral
    client_mistral = Mistral(api_key=mistral_api_key)
    print("✅ Mistral AI Client erfolgreich initialisiert!")
    
    prompt_mistral = "Prüfe folgenden Sachverhalt nach deutschem IT- und Notarrecht: Speicherung von Mandantendaten in einer Ende-zu-Ende verschlüsselten Cloud-Infrastruktur gem. § 203 StGB und DSGVO."
    
    chat_response = client_mistral.chat.complete(
        model="mistral-large-latest",
        messages=[
            {"role": "system", "content": "Du bist ein spezialisierter KI-Jurist und Code-Architekt für deutsches Notar- und Datenschutzrecht."},
            {"role": "user", "content": prompt_mistral}
        ]
    )
    print("\n=== MISTRAL LARGE 2 ANTWORT ===\n")
    print(chat_response.choices[0].message.content)
else:
    print("ℹ️ MISTRAL_API_KEY nicht in Secrets hinterlegt.")
    print("💡 Tipp: Du kannst Mistral Large 2 / Codestral via API nutzen oder Open-Weights Modelle (z.B. Mistral-7B / Codestral) direkt auf der A100 GPU laden.")
```
